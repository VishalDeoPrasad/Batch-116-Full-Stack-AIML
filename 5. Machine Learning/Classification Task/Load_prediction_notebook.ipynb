```python
import pandas as pd
df = pd.read_csv("loan_data.csv")
```

```python
df.info()
```

```python
df.head()
```

```python
df.drop('Loan_ID', axis=1, inplace=True)
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df.head()
```

```python
df.Gender.unique()
```

```python
gender_map = {
    'Male':1, 'Female':0
}
df.Gender = df.Gender.map(gender_map)
```

```python
df.head()
```

```python
df.Married.unique()
```

```python
married_map = {
    'Yes':1, 'No':0
}
df.Married = df.Married.map(married_map)
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.Dependents.unique()
```

```python
df.Dependents = df.Dependents.replace("3+", 3).astype(int)
```

```python
df.dtypes
```

```python
df.head()
```

```python
df.Education.unique()
```

```python
education_map = {
    'Graduate':1, 'Not Graduate':0
}
df.Education = df.Education.map(education_map)
```

```python
df.head()
```

```python
df.Self_Employed.unique()
```

```python
self_employed_map = {
    'No':0, 'Yes':1
}
df.Self_Employed = df.Self_Employed.map(self_employed_map)
```

```python
df.head()
```

```python
df.Property_Area.unique()
```

```python
property_area_map = {
    'Rural':0, 'Urban':2, 'Semiurban':1
}
df.Property_Area = df.Property_Area.map(property_area_map)
```

```python
df.head()
```

```python
df.Loan_Status.unique()
```

```python
load_status_map = {
    'N':0, 'Y':1
}
df.Loan_Status = df.Loan_Status.map(load_status_map)
```

```python
df.head()
```

```python
df.dtypes
```

```python
x = df.drop('Loan_Status', axis=1)
y = df.Loan_Status
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.10)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

model = LogisticRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.svm import SVC

model = SVC()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

model = GradientBoostingClassifier()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy_score(y_test, y_pred)
```
