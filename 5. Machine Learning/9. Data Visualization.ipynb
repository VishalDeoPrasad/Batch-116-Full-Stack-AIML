```python
df = pd.read_csv("Cars24 data.csv")
```

```python
df.head()
```

```python
df.fuel.value_counts().plot(kind='bar')
```

```python
sns.countplot(data=df, x="fuel")
```

```python
df.groupby('fuel')['selling_price'].mean().plot(kind='bar')
```

```python
df.columns
```

```python
sns.scatterplot(data=df, x='km_driven', y='selling_price')
```

```python
sns.boxplot(data=df, x="km_driven")
```

```python
sns.histplot(data=df, x="km_driven")
```

```python
q1 = df.km_driven.quantile(0.25)
q3 = df.km_driven.quantile(0.75)
IQR = q3 - q1
IQR
lower = q1 - 1.5*IQR
upper = q3 + 1.5*IQR

df = df[(df.km_driven >= lower) & (df.km_driven <= upper)]
```

```python
sns.boxplot(data=df, x="km_driven")
```

```python
sns.histplot(data=df, x="km_driven")
```

```python
sns.scatterplot(data=df, x='km_driven', y='selling_price')
```
