### 1. Line Chart using Matplotlib


```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
```

```python
import matplotlib.pyplot as plt
plt.plot(x,y)
plt.title("Line plot");
```

```python
years = [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017]
sales = [24000, 35000, 45000, 47000, 49000, 51000, 55000, 56000]
```

```python
plt.figure(figsize=(8,5))
plt.plot(years,sales, color="green", linestyle='--', marker="o")
plt.title("years vs Sales");
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
years = [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017]
sales = [24000, 15000, 45000, 67000, 49000, 51000, 55000, 56000]
```

```python
plt.plot(years,sales)
```

### 2. Bar Chart

```python
size = ["S", "M", "L", "XL"]
qty = [14, 17, 6, 20]
```

```python
plt.bar(size, qty)
plt.title("T-shirt Size vs Qty")
plt.xlabel("T-shirt Size")
plt.ylabel("Qty")
```

```python
plt.barh(size, qty)
```

### 5. Pie Chart

```python
size = ["S", "M", "L", "XL"]
qty = [14, 17, 6, 20]
```

```python
plt.pie(qty, labels=size, autopct='%1.2f');
plt.title("This is demo");
```

### Subplot

```python
size = ["S", "M", "L", "XL"]
qty = [14, 17, 6, 20]
```

```python
plt.subplot(3,1,1)
plt.bar(size, qty);
plt.title("Plot 1")

plt.subplot(3,1,2)
plt.pie(qty, labels=size, autopct='%1.2f');
plt.title("Plot 2")

plt.subplot(3,1,3)
plt.hist(age);
plt.title("Plot 3")
```
