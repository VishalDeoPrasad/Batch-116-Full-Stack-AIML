#### 2. Bar Chart

```python
import pandas as pd
data = {
    "Product":["Phone","Laptop","Tablet","TV"],
    "Sales" : [50000, 60000, 20000, 30000]
}

df = pd.DataFrame(data)
df
```

```python
px.bar(df, x="Product", y="Sales", color="Product")
```

```python
px.bar(df, x="Sales", y="Product", color="Product",
       orientation="h")
```

#### 3. Scatter Plot

```python
df = px.data.tips()
df
```

```python
px.scatter(df, x="total_bill", y="tip")
```

#### 4. Pie Chart

```python
data = {
    "lang":['Python', 'C', 'C++', 'Java', 'Go'],
    "users" :[50, 30, 40, 45, 15]
}
df = pd.DataFrame(data)
df
```

```python
px.pie(df, names='lang', values="users")
```

#### 5. Donut chart

```python
px.pie(df, names='lang', values="users", 
       hole=0.4)
```
