```python
from sklearn.preprocessing import LabelEncoder
job_title_le = LabelEncoder()
df.job_title = job_title_le.fit_transform(df.job_title)
```

```python
df.company_size.unique()
```

```python
company_size_map = {
    'Small':1, 
    'Medium':2, 
    'Large':3, 
    'Startup':0
}
df.company_size = df.company_size.map(company_size_map)
```

```python
industry_le = LabelEncoder()
df.industry = industry_le.fit_transform(df.industry)
```

```python
df.location.unique()
```

```python
location_le = LabelEncoder()
df.location = location_le.fit_transform(df.location)
```

```python
df.remote_work.unique()
```

```python
remote_work_map = {
    'Yes': 1, 'No':0
}
df.remote_work = df.remote_work.map(remote_work_map)
```

```python
df.head()
```

```python
df.dtypes
```

```python
x = df.drop('salary',axis=1)
y = df.salary
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.10)
```

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
r2_score(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
mean_absolute_error(y_test, y_pred)
```
