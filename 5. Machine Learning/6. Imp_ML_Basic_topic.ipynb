#### <b>1.  Correlation Analysis and Heatmaps</b>

```python
import pandas as pd

# Car = 1, Bike = 2, Cycle = 3
# Red = 1, blue = 2, Green = 3
data = {
    'vehicle':[1, 2, 2, 1, 3],
    'room_color': [1, 1, 1, 3, 1],
    'sleep_hours' : [6, 7, 8, 3, 1],
    'study_hours' : [4, 5, 6, 2, 1],
    'play_hours' : [2, 1, 0.5, 6, 8],
    'exam_marks' : [60, 75, 86, 45, 30]
}
df = pd.DataFrame(data)
df
```

```python
df.corr()
```

```python
import seaborn as sns
sns.heatmap(df.corr(), annot=True)
```

```python
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
```

#### <b>2. Categorical Data Analysis</b>

```python
import pandas as pd

data = {
    "EmployeeID": [101,102,103,104,105],
    "Age": [25,32,29,41,36],
    "Gender": ["Male","Female","Male","Female","Male"],
    "Department": ["IT","HR","Finance","IT","Marketing"],
    "Salary": [50000,62000,58000,75000,54000,],
    "City": ["Delhi","Mumbai","Bangalore","Hyderabad","Chennai"],
    "Country": ["India","India","India","India","India"],
    "Experience": [2,6,4,10,7],
    "Education": ["Bachelors","Masters","Bachelors","PhD","Masters"],
}

df = pd.DataFrame(data)
df
```

```python
df.Gender.value_counts()
```

```python
df.Department.value_counts()
```

```python
df.Country.value_counts()
```
