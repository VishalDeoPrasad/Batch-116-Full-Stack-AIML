```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns

df1 = sns.load_dataset("flights")
df1.head(3)
```

#### 1. Line Chart

```python
sns.lineplot(x="year", y="passengers", data=df1)
plt.title("This is my flights trends")
```

#### 2. Bar Chart

```python
df2 = sns.load_dataset("tips")
df2.head()
```

```python
sns.barplot(x='day', y='total_bill', data=df2)
```

#### 3. Scatter Plot

```python
df3 = sns.load_dataset("iris")
df3.head()
```

```python
sns.scatterplot(x="sepal_length", y='petal_length', data=df3)
```

#### 4. Histogram

```python
sns.histplot(x='sepal_length', data=df3)
```

#### 5. Count Plot

```python
df2.head()
```

```python
sns.countplot(x="day", data=df2)
```
