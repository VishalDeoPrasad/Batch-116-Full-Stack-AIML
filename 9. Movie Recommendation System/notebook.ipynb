```python
import pandas as pd
df = pd.read_csv("movie_dataset.csv")
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['movie_desc'] = df.overview + df.genre
```

```python
df.head()
```

```python
df = df[['title', 'movie_desc']]
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
vector = cv.fit_transform(df.movie_desc)
```

```python
vector[0].toarray()
```

```python
df.title
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
similarity
```

```python
simi_df = pd.DataFrame(similarity, index=df.title, columns=df.title)
```

```python
simi_df
```

```python
def recommend(movie_name):
    score = simi_df[movie_name]
    result = score.sort_values(ascending=False).head(10)
    return list(result.index)
```

```python
recommend('Iron Man')
```

```python
recommend("Spider-Man")
```

```python
list(recommend('Predator').index)
```
