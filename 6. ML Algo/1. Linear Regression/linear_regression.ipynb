```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Hours" : [1,2,3,4,5,6],
    "Marks" : [40, 50, 60, 65, 75, 90]
}
df = pd.DataFrame(data)
df
```

```python
x = df.drop('Marks', axis=1)
y = df.Marks
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x, y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x, model.predict(x), color='red')
```

```python
data = {
    "Experience" : [1, 2, 3, 4, 5],
    "Salary" : [30000, 40000, 45000, 50000, 60000]
}
df = pd.DataFrame(data)
df
```
