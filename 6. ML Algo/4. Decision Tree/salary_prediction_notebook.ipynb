```python
import pandas as pd
```

```python
df = pd.read_csv("salary_data.csv")
```

```python
df.gender.value_counts()
```

```python
df.gender.unique()
```

```python
gender_map = {
    'Male':1, 'Female':0
}
df.gender = df.gender.map(gender_map)
```

```python
df.head()
```

```python
df.education_level.unique()
```

```python
eduction_map = {
    "Bachelor's":0, "Master's":1, 'PhD':3, 'MBA':2
}
df.education_level = df.education_level.map(eduction_map)
```

```python
df.head()
```

```python
df.job_title.unique()
```

```python
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
df.job_title = le1.fit_transform(df.job_title)
```

```python
df.head()
```

```python
df.company_size.unique()
```

```python
company_map = {
    'Small':1, 'Medium':2, 'Large':3, 'Startup':0
}
df.company_size = df.company_size.map(company_map)
```

```python
df.head()
```

```python
df.industry.unique()
```

```python
le2 = LabelEncoder()
df.industry = le2.fit_transform(df.industry)
```

```python
df.head()
```

```python
le3 = LabelEncoder()
df.location = le3.fit_transform(df.location)
```

```python
df.head()
```

```python
df.remote_work.unique()
```

```python
remote_map = {
    'Yes':1, 'No':0
}
df.remote_work = df.remote_work.map(remote_map)
```

```python
df.head()
```

```python
x = df.drop('salary', axis=1)
y = df.salary
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
model.fit(x,y)
```

```python
df.head(0)
```

```python
new_emp = [[32, 1, 1, 10, 6, 1, 6, 5, 0]]
model.predict(new_emp)
```
