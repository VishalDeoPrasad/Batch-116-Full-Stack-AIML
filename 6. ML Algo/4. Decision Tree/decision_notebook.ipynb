```python
import pandas as pd
```

```python
# buy=1, not boy=0 
# low=0, medium=1, High=2
data = {
    "age": [22, 35, 29, 40],
    "income": [0, 2, 1, 2],
    "buy": [0, 1, 1, 1]
}
df = pd.DataFrame(data)
df
```

```python
x = df.drop('buy', axis=1)
y = df.buy
```

```python
x
```

```python
y
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(x, y)
```

```python
model.predict([[23, 2]]) # age=23, income=high
```

```python
model.predict([[23, 0]]) #age=23, income=low
```

```python
from sklearn.tree import plot_tree

plot_tree(model)
```
