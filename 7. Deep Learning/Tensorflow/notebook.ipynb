#### 👩‍🎓Predict Student Performance Based on Study Hours

##### Step 1: Create Dataset

```python
import numpy as np
study_hours = np.array([1,2,3,4,5])
score = np.array([40,50,60,70,80])
```

```python
study_hours, score
```

##### Step 2: Build Model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input, Dropout, BatchNormalization

model = Sequential([
    Input(shape=(1,)),
    Dense(10, activation='relu'),
    Dropout(0.5),
    Dense(1)
])
```

```python
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
model.fit(study_hours, score, epochs=500)
```

```python
model.predict(np.array([4]))
```

#### Deep Neural Network

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input, Dropout, BatchNormalization

model = Sequential([
    Input(shape=(1,)),
    Dense(64, activation='relu'),
    BatchNormalization(),
    Dense(32, activation='relu'),
    Dropout(0.5),
    Dense(1)
])
```

```python
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
model.fit(study_hours, score, epochs=500)
```

```python
model.predict(np.array([2]))
```
