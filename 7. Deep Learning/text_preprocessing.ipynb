```python
text = "I am Learning Deep Learning!@!.=-346"
```

```python
text = text.lower()
```

```python
import re

text = re.sub(r'[^a-zA-Z\s]', '', text)
text
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)
tokens
```

```python
from nltk.corpus import stopwords

text = [word for word in tokens if word not in stopwords.words('english')]
text
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
[stemmer.stem(word) for word in text]
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studies"]
[stemmer.stem(word) for word in words]
```

```python
from nltk.stem import WordNetLemmatizer
lemma = WordNetLemmatizer()

words = ["playing", "played", "plays", "studies"]
[lemma.lemmatize(word, pos='v') for word in words]
```
