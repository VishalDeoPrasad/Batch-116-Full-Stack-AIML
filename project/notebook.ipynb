```python
import pandas as pd
import random

# Number of rows
num_rows = 50

# Sample names
names = [
    "Aarav","Vihaan","Arjun","Sai","Krishna","Rahul","Rohan","Aditya","Kiran","Varun",
    "Ananya","Aditi","Pooja","Sneha","Kavya","Divya","Meera","Isha","Nisha","Priya"
]

grades = ["A", "B", "C", "D"]

data = []

for i in range(1, num_rows + 1):
    math = random.randint(40, 100)
    science = random.randint(40, 100)
    english = random.randint(40, 100)
    
    avg_score = (math + science + english) / 3
    passed = "Yes" if avg_score >= 50 else "No"
    
    row = {
        "Student_ID": i,
        "Name": random.choice(names),
        "Age": random.randint(15, 20),
        "Gender": random.choice(["Male", "Female"]),
        "Grade": random.choice(grades),
        "Math_Score": math,
        "Science_Score": science,
        "English_Score": english,
        "Attendance_Percentage": random.randint(60, 100),
        "Passed": passed
    }
    
    data.append(row)

# Create DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("student_dataset.csv", index=False)

print("CSV file 'student_dataset.csv' created successfully!")
```
