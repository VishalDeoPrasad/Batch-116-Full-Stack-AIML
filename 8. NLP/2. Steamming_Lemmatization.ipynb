```python
words = ['playing', 'running', 'walked', 'jumped',
         'cats', 'dogs', 'books', 'quickly', 'slowly',
         'education', 'creation', 'studies']
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()

for word in words:
    print(ps.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer
wl = WordNetLemmatizer()

for word in words:
    print(wl.lemmatize(word))
```
